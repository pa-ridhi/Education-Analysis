```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\parid\OneDrive\Documents\Education_Analysis\nirf_university_rankings.csv")

print("Shape:", df.shape)
print("\nColumns:",df.columns.tolist())
df.head()
```

```python
top_states= df['State'].value_counts().head(10)
print(top_states)

top_states.plot(kind='bar', color='royalblue', figsize=(10, 5))
plt.title('Top 10 States with Most University in NIRF Ranking')
plt.xlabel('State')
plt.ylabel('Number of Universities')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
top_universities=df.nlargest(10,'Score')[['Name', 'State', 'Score']]
print(top_universities)

df.nlargest(10, 'Score').plot(kind='bar', x='Name', y='Score', color='green', figsize=(12,5))
plt.title('Top 10 University by NIRF Score')
plt.xlabel('University')
plt.ylabel('Score')
plt.xticks(rotation= 45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\parid\OneDrive\Documents\Education_Analysis\nirf_university_rankings.csv")

print("Data loaded successfully!")
print("Shape:", df.shape)
```

```python
avg_score= df.groupby('State')['Score'].mean().sort_values(ascending=False).head(10)
print(avg_score)

avg_score.plot(kind='bar', color='coral', figsize=(10,5))
plt.title('Top 10 States by Average University Score')
plt.xlabel('State')
plt.ylabel('Average Score')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
top_cities= df['City'].value_counts().head(10)
print(top_cities)

top_cities.plot(kind='bar', color='purple', figsize=(10,5))
plt.title('Top 10 Cities with Most Universities in NIRF rankng')
plt.xlabel('City')
plt.ylabel('Number of Universities')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
df['Score'].plot(kind='hist', bins=20, color='teal', figsize=(10,5))
plt.title('Distribution of NIRF Scores')
plt.xlabel('Score')
plt.ylabel('Number of Universities')
plt.tight_layout()
plt.show()

print("Average Score:", df['Score'].mean().round(2))
print("Highest Score:", df['Score'].max())
print("Lowerst Score:", df['Score'].min())
```

```python
top10= df.nlargest(10, 'Score')[['Name', 'State', 'City', 'Score', 'Rank']]
print(top10.to_string())

top10.plot(kind='bar', x='Name', y='Score', color='gold', figsize=(12,5))
plt.title('Top 10 Universities by NIRF Score 2024')
plt.xlabel('University')
plt.ylabel('Score')
plt.xticks(rotation= 45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\parid\OneDrive\Documents\Education_Analysis\nirf_university_rankings.csv")

top_states = df['State'].value_counts().head(10)
top_cities = df['City'].value_counts().head(10)
avg_score = df.groupby('State')['Score'].mean().sort_values(ascending=False).head(10)
top10 = df.nlargest(10, 'Score')[['Name', 'State', 'City', 'Score', 'Rank']]

with pd.ExcelWriter('Education_Analysis_Results.xlsx') as writer:
    top_states.to_excel(writer, sheet_name='States')
    top_cities.to_excel(writer, sheet_name='Cities')
    avg_score.to_excel(writer, sheet_name='Avg Score by State')
    top10.to_excel(writer, sheet_name='Top 10 Universities')

print("Excel file saved successfully!")
```
